```python
from sklearn.linear_model import LinearRegression
#house size
X=[[1000],[1200],[1500],[1800]]
#house price in lakhs
Y=[40,50,65,80]
model=LinearRegression()
model.fit(X,Y)
prediction=model.predict([[1600]])
print("Predicted Prize",prediction[0],"Lakh")
```

```python
from sklearn.linear_model import LinearRegression
X = [[8.77],[8.18]]
Y = [8.18,8.77]
model = LinearRegression()
model.fit(X, Y)
target_sem1 = [[8.77]]
prediction = model.predict(target_sem1)
print(prediction[0])
```

```python
import numpy as np
from sklearn.cluster import KMeans
data = np.array([
    [2, 3],
    [3, 4],
    [10, 12],
    [11, 13]
])
model = KMeans(n_clusters=2, random_state=0, n_init="auto").fit(data)
print("Labels:", model.labels_)
print("Centers:", model.cluster_centers_)
```
